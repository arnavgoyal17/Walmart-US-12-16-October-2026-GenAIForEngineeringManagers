```python
# AI vs ML vs DL vs GenAI vs Agentic AI

# https://miro.medium.com/0*RBFoYaZlkPGS2K-u.png

# 1. Artificial Intelligence (AI) — Making machines behave intelligently
# Imagine a Walmart store running low on a popular product.
# A system looks at the current inventory, replenishment rules, expected demand, and delivery lead time, then flags that stock needs to be replenished.
# That is a practical use of AI-supported decision-making.
# AI describes the broad capability, not the particular technology underneath it.

# 2. Machine Learning (ML) — Learning from historical data
# Predicting product demand
# Suppose Walmart wants to predict how many bottles of water a store is likely to sell next Saturday.
# We have historical sales, store location, weather, holidays, promotional activity, and seasonal demand.
# An ML model can learn patterns from that information.
# It might predict that a particular store is likely to sell 1,200 bottles next Saturday.
# ML has predicted a number. It hasn't written an explanation, generated an advertisement, or placed an inventory order.

# Leadership takeaway: If the problem involves prediction, classification, scoring, or detecting patterns in structured data, traditional ML is often worth evaluating before considering an LLM.

# 3. Deep Learning (DL) — Learning more complex patterns
# Identifying products using cameras
# Imagine a camera-assisted checkout system.
# A customer places a product near the camera, and the system needs to recognize what it is.
# A Deep Learning computer vision model could analyze the image and identify the product.
# This is fundamentally different from demand forecasting.
# One problem involves predicting future demand from historical data.
# The other involves interpreting complex visual information.
# And neither of these necessarily requires Generative AI.

# The right technology is the one that solves the problem with acceptable accuracy, latency, cost, and operational complexity.

# 4. Generative AI (GenAI) — Creating new content
# Generating a convincing answer is not the same as verifying that the answer is correct.
# eg: Customer review intelligence
# Imagine Walmart receives thousands of customer reviews about a particular product.
# An Engineering Manager is asked to build a solution that helps the category management team understand those reviews.
# A traditional ML model might classify the reviews as positive, negative, or neutral.
# Generative AI can go further.
# It could produce a summary like:
#     'Customers generally appreciate the product's price and performance, but a recurring complaint concerns packaging damage during delivery.'
# It could then draft a business-friendly report with suggested areas to investigate.

# 4. Agentic AI — Moving from generating answers to completing tasks
# An agentic system can examine the situation, decide which available tools to use, take actions, observe results, and continue until it reaches its goal or needs human intervention.
# We can decide exactly how much freedom it gets.
# Walmart example — Delayed delivery investigation
# Let's compare how different technologies could contribute to the same business problem.

# Scenario: A customer reports that an online order has not arrived.

# 1. ML predicts a delay
# A predictive model estimates that an order is at risk of missing its delivery commitment.

# 2. GenAI explains the situation
# An LLM produces a clear, customer-friendly explanation from verified order information.

# 3. Agentic AI coordinates resolution
# An agent checks authorized systems, retrieves shipment events, determines the next approved workflow step, creates a support ticket, and prepares a notification.

# 4. Human approval protects higher-risk actions
# If compensation or another sensitive action is required, the agent pauses until an authorized employee approves it.
```

## The Engineering Manager's comparison

| Technology | Primary purpose                           | Walmart example                                | What to evaluate                                  |
| ---------- | ----------------------------------------- | ---------------------------------------------- | ------------------------------------------------- |
| AI         | Intelligent behavior or decision-making   | Rule-based inventory alerts                    | Correctness, business logic                       |
| ML         | Learn patterns and make predictions       | Sales and demand forecasting                   | Prediction accuracy, drift                        |
| DL         | Learn complex representations             | Computer vision for product recognition        | Data quality, compute, latency                    |
| GenAI      | Generate and transform content            | Summarize customer reviews                     | Faithfulness, relevance, cost                     |
| Agentic AI | Pursue goals through tool use and actions | Investigate delivery issues and create tickets | Task success, permissions, reliability, oversight |

```python
# AI is the broad field of making machines perform intelligent tasks.

# Machine Learning learns patterns from data and often uses them to make predictions.

# Deep Learning uses multilayer neural networks to learn complex patterns.

# Generative AI creates and transforms content such as text, images, and code.

# Agentic AI uses goals, models, tools, and feedback to coordinate actions, subject to the controls we design.
```

```python
# Biggest takeaway for everyone managing engineering tea:
# Your job is not to introduce the most sophisticated AI solution.

# Your job is to select the simplest architecture that reliably delivers the required business outcome, while meeting your expectations for cost, security, scalability, and governance.
```

```python
# What Generative AI is — what a large language model is actually doing when it answers you

# When you open ChatGPT and type a question, what exactly happens?
# Let's say I ask:
#     'Why are grocery sales increasing during weekends?'
# Within a few seconds, ChatGPT gives me a beautifully structured answer. It talks about shopping behavior, families purchasing groceries together, promotions, and consumer trends.
# Sometimes, the explanation is so convincing that it feels as though we're speaking to an experienced retail consultant.
# But here's my question to all of you:
# Does ChatGPT actually understand Walmart's grocery business? Has it examined Walmart's latest sales numbers? Or is something else happening?
# By default, a language model doesn't automatically access Walmart's internal sales systems, inventory databases, or current business reports.
# Instead, it generates an answer using patterns it learned during training, combined with the information we provide in the conversation and any tools or external sources made available to it.
```

```python
# What is a Large Language Model (LLM)?
# A large language model is a type of generative AI that has been trained on vast amounts of text data. It learns patterns, structures, and relationships in the language, enabling it to generate coherent and contextually relevant responses to prompts. LLMs can perform various tasks such as answering questions, summarizing text, translating languages, and even generating creative content. However, they do not possess true understanding or consciousness; they rely on statistical patterns learned from their training data to produce outputs.

# An LLM learns statistical patterns from vast amounts of training data.

# https://developers.openai.com/api/docs/pricing

# https://platform.openai.com/tokenizer
```

```python
# Why can the same question produce different answers?

# One major reason is how the model selects tokens.
# A model assigns probabilities to possible continuations.
# Depending on the decoding settings, it may choose different reasonable continuations on different runs.

# For a production system that must return a fixed JSON structure or apply an exact refund policy, uncontrolled variation may be unacceptable.
# And that's why we need structured outputs, validation, deterministic business rules, and appropriate generation settings.
```

```python
from openai import OpenAI
client = OpenAI(api_key=openai.api_key)

completion = client.chat.completions.create(
    model="gpt-5.5",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write a poem about the beauty of nature."}
    ]
)

# print(completion)
print(completion.choices[0].message.content)

# There are 3 types of messages:
# 1. user - I am the user or whosoever is asking a question to the GPT model is called as the user.
# 2. assistant - GPT model is the assistant.
# 3. system - this message is the aura or the persona, which you want the GPT model to adopt while answering the question. It is optional but it can help in getting better answers from the model.
```

```python
# Tokenization Demo
import tiktoken
enc = tiktoken.encoding_for_model("gpt-4o")

texts = [
    "Walmart",
    "Walmart processes 2.3 million customer transactions per hour.",
    "Supply chain optimization using LangGraph and LLM-as-judge evaluation.",
    "SKU-7829341-B",   # A product identifier — watch how this is split
    "def run_agent(state: AgentState) -> AgentState:",  # Code
]

print(f"Tokenizer: {enc.name}")
print(f"Vocabulary size: {enc.n_vocab:,} tokens")
print()
print(f"{'Text':<55} | {'Tokens':>6} | {'Token IDs (first 8)'}")
print("-" * 90)

for text in texts:
    token_ids = enc.encode(text)
    token_words = [enc.decode([t]) for t in token_ids]
    preview_ids = str(token_ids[:8]) + ("..." if len(token_ids) > 8 else "")
    print(f"{text[:54]:<55} | {len(token_ids):>6} | {preview_ids}")

print()
print("Key insight: 'SKU-7829341-B' is split into multiple tokens.")
print("Structured identifiers cost more tokens than natural text.")
```
